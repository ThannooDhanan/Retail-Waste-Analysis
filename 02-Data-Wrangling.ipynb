# Data Wrangling

I will now format the data. Especially remapping the TERMINATRED column into something that makes more sense.

```python
!pip install ipython-sql
!pip install seaborn
import pandas as pd
import sqlite3

from pathlib import Path

data_dir = Path.cwd()

df = pd.read_csv(data_dir / 'Data' / 'Canada_Food_Waste.csv')

df = df.drop(columns = 'Unnamed: 0')

df.head()
```

```python
TERMINATED_map = {'t' : True}

df['TERMINATED'] = df['TERMINATED'].map(TERMINATED_map).fillna(False)
```

```python
df[df['TERMINATED'] == False]
```

Success! The data frame' TERMINATED column has been remapped!

```python
df['TERMINATED'].value_counts(dropna=False)
```

# Now I want to put the data frame into an SQLite database

```python
conn = sqlite3.connect('Food_Waste.db')
cursor = conn.cursor()

df.to_sql('CA_Food_Waste', conn, if_exists= 'replace', index=False)
```
